```julia
##############################
# RUN MPC
##############################
using JuMP

n_steps = Int(nt/dt)+1     
#Fiks indeksering i profiles
Vprofile = zeros(n_steps,1)  
Ssprofile = zeros(n_steps, 1)
Xbprofile = zeros(n_steps, 1)
Soprofile = zeros(n_steps, 1)
q_inprofile = zeros(n_steps,1)
q_outprofile = zeros(n_steps,1)

V0 = 1000
Ss0 = 2.727
Xb0 = 18.357
So0 = 9.75

Vprofile[1] = V0
Ssprofile[1] = Ss0
Xbprofile[1] = Xb0
Soprofile[1] = So0
q_inprofile[1] = 100.0
q_outprofile[1] = 100.0



for i = 2:n_steps

    # CHANGES TO CONSTRAINTS AND DISTURBANCE
    qmaxin = qmaxin_vec[i]
    qmaxout = qmaxout_vec[i]

    # FIX INITIAL CONDITIONS IN MODEL AND RUN 
    fix.(mpc_model[:V][1, 1], V0, force=true)
    fix.(mpc_model[:Ss][1, 1], Ss0, force=true)
    fix.(mpc_model[:Xb][1, 1], Xb0, force=true)
    fix.(mpc_model[:So][1, 1], So0, force=true)

    # FIX DISTURBANCES IN MODEL
    fix.(mpc_model[:qmaxin], qmaxin, force=true)
    fix.(mpc_model[:qmaxout], qmaxout, force=true)

    # SOLVE MODEL 
    optimize!(mpc_model)
    status = JuMP.termination_status(mpc_model)

    if status != MOI.LOCALLY_SOLVED && status != MOI.OPTIMAL
        println("Model did not converge to global or local optima")
    else
        println("Model solved successfully")
    end
    
    println(status)
    

    # REASSIGN INITIAL CONDITIONS 
    # SAVE RESULTS FROM 2ND TIME STEP (FIRST TIME STEP IS INITIAL CONDITION)
    V0 = value(mpc_model[:V][1, 2]) 
    Ss0 = value(mpc_model[:Ss][1, 2])
    Xb0 = value(mpc_model[:Xb][1, 2])
    So0 = value(mpc_model[:So][1, 2])

    
    Vprofile[i] = V0 
    Ssprofile[i] = Ss0 
    Xbprofile[i] = Xb0 
    Soprofile[i] = So0 
    q_inprofile[i] = value(mpc_model[:q_in][1])
    q_outprofile[i] = value(mpc_model[:q_out][1])


end 
```
